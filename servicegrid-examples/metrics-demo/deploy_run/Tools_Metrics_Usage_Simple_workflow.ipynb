# [ServiceGrid](https://github.com/opencyber-space/Servicegr.id)
```
                  ServiceGrid
                    __|____________
                   |              |
          Tools Bank             Functions Bank
         |- Python Code(Type-1)   |- Statefull Functions(Using K8s Deployment)(Type-3)
         |- Binary Code           |- Stateless Jobs(Using K8s Jobs)(Type-2)
```
**Note**:
- Tools will be run within Agent Code i.e it shares CPU Cycles, RAM, Disk of the Agent Pod
- Functions will be run outside of Agent Code i.e CPU Cycles, RAM and Disk are not from Agent Pod.

| Type | Description | Functions Lifecycle | Communication | Scaling | Fault Tolerance |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Type 1**<br>Functions loaded and executed inside an application | Executed within the application's own process using the Local Functions Evaluation SDK | Managed by the application; can be loaded/unloaded anytime | In-process memory | Handled by the application (manual process/thread management) | Depends on the application |
| **Type 2**<br>Functions deployed as a job on remote clusters | One-time stateless Functions execution on a remote Kubernetes cluster | Stateless; runs as a Kubernetes Job and terminates upon completion | No live communication; application waits for results | Not applicable | Handled by Kubernetes |
| **Type 3**<br>Functions deployed as a stateful online function | Stateful Functions running as long-lived Kubernetes pods, exposed via REST API | Stateful; initialized once and runs as a pod | REST API over HTTP | Autoscaling rules configurable at deployment | Handled by Kubernetes |


## 1. Directory Structure
Understanding the files in this demonstration:
* **`nodes/`**: Contains the Python code for each agent (`agent_tool_metrics_demo.py`).
* **`spec/`**: Contains agent registration JSON specifications (`agent_tool_metrics_demo.json`, etc.) and bash scripts for agent-level operations (registering, deploying, removing, unregistering).
* **`deploy_run/`**: Contains the `workflow_spec-tools.json` (which defines the data flow between the agents), the Streamlit dashboard code (`streamlit_app.py`, `structure`), and the bash scripts for workflow orchestration and inference.
* **`tools/`**: Containes files related to codes of the tools, its registration spec files, deployment of tools etc
